```python
def autenticar():

    url = f"{URL}/Login/Autenticar"

    resposta = sessao.post(
        url,
        params={"token": TOKEN}
    )

    if resposta.status_code == 200 and resposta.json() == True:

        print("Autenticação autorizada!")
        return True

    print("Falha na autenticação.")
    print(resposta.text)

    return False
```

```python
def buscar_linhas(busca):
    url = f"{URL}/Linha/Buscar"

    resposta = sessao.get(url, params={"termosBusca": busca})

    if resposta.status_code == 200:
        linhas = resposta.json()
        return linhas

    print("Falha ao buscar linhas.")
    return resposta.json()
```
